```python
import pandas as pd

treasure_df = treasure_df = pd.read_csv('../data/heritage_exp.csv')
treasure_df = " ".join(treasure_df["exp"].dropna())
treasure_df = treasure_df.replace('\r\n\r\n', ' ')
treasure_df
```

```python
from nltk import sent_tokenize
import re

# 괄호안 내용 제거
sentences = sent_tokenize(treasure_df)
sentences = [re.sub(r'\(.*?\)', '', s) for s in sentences]
print(sentences)
```

```python
df_sentences = pd.DataFrame({"sentence": sentences})
df_sentences.to_csv("../data/heritage_sent.csv", index=False)
```

```python
import pandas as pd

# 문장이 '다.'로 안끝날 경우 삭제
treasure_df= pd.read_csv('../data/heritage_sent.csv')
for i in range(len(treasure_df['sentence'])):
    sent = treasure_df['sentence'].loc[i]
    if len(sent)<2:
        treasure_df = treasure_df.drop(i)
    else:
        end_sent=sent[-2]
        if end_sent!='다':
            treasure_df = treasure_df.drop(i)       
        
treasure_df.to_csv('../data/heritage_sent.csv', index=False)
```

### 한자 추출

```python
treasure_df= pd.read_csv('../data/heritage_par.csv')
import re
kor = []
china = []
for i in range(len(treasure_df['sentence'])):
    text = treasure_df['sentence'].loc[i]
    matches = re.findall(r'([가-힣]+)\s*\(([\u4e00-\u9fff]+)\)', text)
    if (len(matches)!= 0):
        kor.append([p[0] for p in matches])
        china.append([p[1] for p in matches])
kor_word = [word2 for word in kor for word2 in word]
china_word = [word2 for word in china for word2 in word]

print(kor_word)
print(china_word)
df = pd.DataFrame({'kor': kor_word, 'hanja': china_word})
df.to_csv('../data/heritage_hanja.csv', index=False, encoding='utf-8-sig')
```

### 불용어 제거, 어간 처리

```python
from konlpy.tag import Okt

treasure_df= pd.read_csv('../data/heritage_sent.csv')
text = " ".join(treasure_df['sentence'].astype(str).tolist())
okt = Okt()
tokens = okt.nouns(text)
```

```python
# 불용어 파일 읽기
def load_stopwords(filepath):
    with open(filepath, 'r', encoding='UTF-8') as f:
        return [line.strip() for line in f if line.strip()]
                
ko_stopwords = load_stopwords('ko_stopwords.txt')

cleaned_tokens = [token for token in tokens if token not in ko_stopwords]
cleaned_tokens = [token for token in cleaned_tokens if len(token) > 1]
cleaned_tokens = [token for token in cleaned_tokens if len(token) and re.match(r'[가-힣]', token)]

df = pd.DataFrame(cleaned_tokens, columns=['word'])
df = df.drop_duplicates()
df.to_csv('../data/heritage_token.csv', index=False)
```
